```python
def noisepsd(dat, fs, rms=True):
    dat = dat - np.mean(dat)
    datfft = rfft(dat)
    fftfreq = rfftfreq(len(dat), 1/fs)
    if rms:
        scale = 2
    else:
        scale = np.sqrt(2)

    inoise_psd = 1*((len(dat)/fs)*(1*(1/(25813/3))*((scale/len(dat)*np.abs(datfft)))/1)**2)# A^2/Hz

    return fftfreq, inoise_psd
```
